```python
acct, cont = "ecommercelake1", "raw"
sas = "<SAS_TOKEN>" 
spark.conf.set(f"fs.azure.sas.{cont}.{acct}.blob.core.windows.net", sas)
base = f"wasbs://{cont}@{acct}.blob.core.windows.net"

for f in notebookutils.fs.ls(base):
    print(f.name, f.isDir)
```

```python
tables = ["customers", "products", "order", "order_items", "payments"]

for t in tables:
    df = (spark.read.option("header", True).option("inferSchema", True)
          .option("recursiveFileLookup", True)
          .csv(f"{base}/{t}/"))
    df.write.mode("overwrite").format("delta").saveAsTable(f"bronze_{t}")
    print(t, df.count())
```
